## 1. Array-Based Stack

A stack follows the **LIFO** (Last In, First Out) principle.  
This implementation stores elements in a Python list and supports the standard stack operations.


```python
class Empty(Exception):
    pass


class ArrayStack:
    def __init__(self):
        self.data = []

    def __len__(self):
        return len(self.data)

    def push(self, x):
        self.data.append(x)

    def pop(self):
        if self.is_empty():
            raise Empty("nothing to remove!")
        return self.data.pop()

    def top(self):
        if self.is_empty():
            raise Empty("nothing to show!")
        return self.data[-1]

    def is_empty(self):
        return len(self.data) == 0
```

## 2. Balanced Parentheses

A stack can be used to verify whether parentheses, brackets, and braces are correctly matched and nested.


```python
def is_matched(expr):
    stack = ArrayStack()
    close_to_open = {
        ")": "(",
        "}": "{",
        "]": "["
    }

    for char in expr:
        if char in close_to_open.values():
            stack.push(char)
        elif char in close_to_open:
            if stack.is_empty():
                return False
            if stack.top() != close_to_open[char]:
                return False
            stack.pop()

    return stack.is_empty()


print(is_matched("(df(d))"))
print(is_matched("([{}])"))
print(is_matched("([)]"))
```

## 3. Circular Array Queue

A queue follows the **FIFO** (First In, First Out) principle.

This implementation uses a circular array so that dequeued positions can be reused. When the queue reaches capacity, the internal array is resized.


```python
class ArrayQueue:
    def __init__(self, capacity):
        self.capacity = capacity
        self.data = [None] * self.capacity
        self.front = 0
        self.size = 0

    def __len__(self):
        return self.size

    def is_empty(self):
        return self.size == 0

    def dequeue(self):
        if self.is_empty():
            raise Empty("nothing to remove")

        answer = self.data[self.front]
        self.data[self.front] = None
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return answer

    def enqueue(self, x):
        if self.size == self.capacity:
            self.resize(len(self.data))

        end = (self.front + self.size) % self.capacity
        self.data[end] = x
        self.size += 1

    def resize(self, length):
        old_data = self.data
        self.capacity = length * 2
        self.data = [None] * self.capacity

        j = self.front
        for i in range(self.size):
            self.data[i] = old_data[j]
            j = (j + 1) % len(old_data)

        self.front = 0
```

## 4. Linked Stack with Middle Removal

This exercise extends the stack idea with a doubly linked list and keeps a reference to the middle node, allowing the middle element to be removed without traversing the entire structure.


```python
class MiddleStack:
    class Node:
        def __init__(self, element):
            self.element = element
            self.prev = None
            self.next = None

    def __init__(self):
        self.head = None
        self.tail = None
        self.middle = None
        self.size = 0

    def push(self, x):
        new_node = self.Node(x)

        if self.size == 0:
            self.head = self.middle = self.tail = new_node
            self.size = 1
            return

        new_node.prev = self.tail
        self.tail.next = new_node
        self.tail = new_node
        self.size += 1

        # For an odd-sized stack, move the middle one step right.
        if self.size % 2 == 1:
            self.middle = self.middle.next

    def pop(self):
        if self.size == 0:
            raise Empty("nothing to remove")

        value = self.tail.element

        if self.size == 1:
            self.head = self.middle = self.tail = None
            self.size = 0
            return value

        old_size = self.size
        self.tail = self.tail.prev
        self.tail.next = None
        self.size -= 1

        # When an odd-sized stack becomes even, move the middle left.
        if old_size % 2 == 1:
            self.middle = self.middle.prev

        return value

    def remove_middle(self):
        if self.size == 0:
            raise Empty("nothing to remove")

        value = self.middle.element

        if self.size == 1:
            self.head = self.middle = self.tail = None
            self.size = 0
            return value

        old_middle = self.middle
        old_size = self.size

        if old_middle.prev:
            old_middle.prev.next = old_middle.next
        if old_middle.next:
            old_middle.next.prev = old_middle.prev

        self.size -= 1

        # Keep the same middle convention used by push/pop.
        if old_size % 2 == 1:
            self.middle = old_middle.prev
        else:
            self.middle = old_middle.next

        return value
```
